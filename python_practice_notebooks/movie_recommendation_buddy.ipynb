```python
import os
from dotenv import load_dotenv
from groq import Groq
load_dotenv()

# List to store the full conversation with the agent and create a memory
conversation = []

#define the client
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

def chat(user_message):
    #add what user said to the memory list
    conversation.append({
        "role": "user",
        "content": user_message
    })
    
    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=conversation,
            temperature=0.2
        )

        reply = response.choices[0].message.content
    except Exception as e:
        return f"Sorry something went wrong: {e}"

    # add the AI's replyto memory too, so it remembers next time
    conversation.append({
        "role": "assistant",
        "content": reply
    })
    return reply
```

```python
chat("I love horror and thrillers, and I can't stand sci-fi movies.")
```

```python
chat("Suggest me a movie to watch tonight")
```

```python
def calculate_margin(revenue,expense):
    profit = revenue - expense
    margin = profit * 100 / revenue
    return margin
```

```python
calculate_margin(60,10)
```

```python
def calculate_profit_and_margin(revenue,expense):
    profit = revenue - expense
    margin = profit * 100 / revenue
    return profit,margin
```

```python
calculate_profit_and_margin(80,30)
```

```python
#type hint
def calculate_profit_and_margin(revenue: float,expense: float) -> float:
    profit = revenue - expense
    margin = profit * 100 / revenue
    return profit,margin
```

```python
def calculate_expense(*args):
    print(args)
    return sum(args)

calculate_expense(2,3,5)
```

```python
def total_expenses(**kargs):
    print(kargs)
    total = 0
    for key,value in kargs.items():
        total += value
    return total

total_expenses(rent=1000, bill=30)
```

```python
#Exercise
final_bill_amount = 1
chai_per_cup_price = 15

def chai_bill(cups, *addons, **discount):
    if cups==0:
        print("User must provide valid number of cups")
        
    final_bill_amount = cups * chai_per_cup_price
    
    if len(addons)>0:
        final_bill_amount = final_bill_amount + 5*len(addons)

    if len(discount)>0:
        discount_percent = discount.get("discount_percent")
        final_bill_amount = (final_bill_amount - ((final_bill_amount * discount_percent)/100))
        
    return final_bill_amount 

print(chai_bill(2))
print(chai_bill(2,"ginger","masala"))
print(chai_bill(4,"biscuit",discount_percent=10))
print(chai_bill(5, "ginger", "masala", "biscuit", discount_percent=10))
```

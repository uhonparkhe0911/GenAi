```python
from datetime import date

class Human:

    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = birth_year

    def get_age(self):
        current_year = date.today().year
        return current_year - self.birth_year

    def introduce(self):
        print(f"Hi, my name is: {self.name}. I am {self.get_age()} years old.")

human1 = Human("Utkarsh", 1995)
human1.introduce()

human2 = Human("Anushka", 1994)
human2.introduce()
```

```python
from pydantic import BaseModel

class Recipe(BaseModel):
    title: str
    calories: int

dal = Recipe(title=120, calories=280)
print(dal.title)
print(dal.calories)
```

```python
class YouTubeChannel:

    def __init__(self, name):
        self.name = name
        self.subscribers = 0
        print(f"Welcome to my YouTube Channel: {self.name}. I have {self.subscribers} subscribers")

    def subscribe(self):
        self.subscribers += 1
        print(f"Thanks for subscribing! Total subscribers now: {self.subscribers}")

    def unsubscribe(self):
        if self.subscribers>0:
            self.subscribers -= 1
            print(f"\nTotal subscribers now: {self.subscribers}")

    def show(self):
        print(f"\n{self.name} has {self.subscribers} subscribers")
            
channel = YouTubeChannel("codebasics")
channel.subscribe()
channel.subscribe()
channel.subscribe()
channel.subscribe()

channel.unsubscribe()
channel.unsubscribe()

channel.show()
```

```python
def safe_chai_bill(cups, price_per_cup):
    total_bill = 0
    try:
        number_of_cups = int(cups)
        price_per_cup = int(price_per_cup)
        print("Number of cups:", number_of_cups)
        print("Price per cup:", price_per_cup)
        if number_of_cups < 1:
            return "Cups must be atleast 1"
        total_bill = number_of_cups * price_per_cup
        return f"Total Bill: {total_bill}"
    except ValueError:
        return "Please enter numbers only"
```

```python
safe_chai_bill(3, 15)
```

```python
safe_chai_bill("3", "15")
```

```python
safe_chai_bill("three", "15")
```

```python
safe_chai_bill(0, "15")
```

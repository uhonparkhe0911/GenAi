```python
def generate_recipe(ingredients, cuisine, diet):
    '''
    Generate a recipe using provided ingredients.
    Cusine can be "Italian", "Mexican", "Indian", "Any" etc.
    Diet can be "Vegetarian", "Vegan", "NonVegetarian", "Jain", "Any" etc
    '''
    prompt = f'''
    Generate one good recipe for me using these ingredients : {','.join(ingredients)}
    Recipe should not be more than 100 words
    Cuisine: {cuisine}
    Diet: {diet}
    '''
    response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.2
)
    return response.choices[0].message.content
```

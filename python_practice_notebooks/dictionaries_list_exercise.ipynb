```python
student_courses = { 
    "Vinit": ("Maths", "Physics"),
    "Anushka": ("Maths", "Chemistry"),
    "Utkarsh": ("Physics", "English"),
    "Koustubh": ("History", "Computer"),
    "Umesh": ("Chemistry", "Computer")
}
all_courses = []
student_enrolled_in_maths = []
unique_courses = []
course_count = {}
students_enrolled_in_computer_maths = []
print("\nTask 1: Print all students and their courses.")
      
for student, courses in student_courses.items():
    print(student, ":", courses)

print("\nTask 2: Find students enrolled in Maths")

for student, courses in student_courses.items():
    if "Maths" in courses:
        student_enrolled_in_maths.append(student)

print("Students enrolled in Maths are:", student_enrolled_in_maths)

print("\nTask 3: Collect all courses into a list")
for courses in student_courses.values():
    all_courses.extend(courses)
print("All the Courses in school are:", all_courses)

print("\nTask 4: Find all unique courses using a set")
unique_courses = set(all_courses)
print("Unique courses in school are:", unique_courses)

print("\nTask 5: Count students enrolled in each course")
for courses in student_courses.values():
    for course in courses:
        if course in course_count:
            course_count[course] += 1
        else:
            course_count[course] = 1

for course,count in course_count.items():
    print(course, ":", count)

print("\nTask 6: Find students enrolled in Computer or Maths")
for student, course in student_courses.items():
    if "Computer" in course or "Maths" in course:
        students_enrolled_in_computer_maths.append(student)

print("Students enrolled in computer are:", students_enrolled_in_computer_maths)
        
```

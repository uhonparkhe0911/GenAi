```python
financials = {
    "Q1" : { "revenue" : 50 , "expense" : 10 },
    "Q2" : { "revenue" : 60 , "expense" : 30 },
    "Q3" : { "revenue" : 80 , "expense" : 20 },
    "Q4" : { "revenue" : -10 , "expense" : 20 },
}

financials

for quarter, data in financials.items():
    if( data["revenue"] < 0 ):
        print(f"Invalid revenue for {quarter}. Skipping")
        continue
        
    profit = data["revenue"] - data["expense"]
    margin = profit * 100 / data["revenue"]
    print(f"{quarter} : {margin:.2f}")
```

```python
#Exercise to calculate average of each student using for loop

students = {
    "Aarav":   [85, 90, 78],
    "Priya":   [72, 68, 75],
    "Rohan":   [45, 52, 48],
    "Sneha":   [95, 92, 98],
    "Manish":  [60, 65, 70],
}

print("=== Student Averages ===")

def calculate_average_and_result(students):
    topper_average = 0
    topper_name = ""
    for student,score in students.items():
        average = 0
        total = 0
        
        for marks in score:
            total = total+marks
        average = (total/300 * 100)

        if(average>=80):
            result = "Topper"
        elif(average>=60 and average<=79):
            result = "Pass"
        else:
            result = "Needs Improvement"

        print(f"'''{student}: {average:.2f} - {result}'''")

        if average>topper_average:
            topper_average = average
            topper_name = student
    print("\n")
    print("=== Class Topper ===")
    print(f"{topper_name} with average {topper_average:.2f}")
        
calculate_average_and_result(students)

        
```
